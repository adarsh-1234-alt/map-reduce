```python
students = [
    "101,Ankit,CSE",
    "102,Riya,IT",
    "103,Rohan,CSE",
    "104,Sunny,DS",
    "105,Aman,CSE",
    "106,Priya,IT",
    "107,Mariam,DS",
    "108,Simran,CSE"
]
```

```python
def mapper(line):
    data = line.strip().split(',')
    department = data[2]
    return (department, 1)

mapped_data = []

for line in students:
    mapped_data.append(mapper(line))

print("Mapper Output:")
print(mapped_data)
```

```python
grouped_data = {}

for department, count in mapped_data:
    if department not in grouped_data:
        grouped_data[department] = []
    grouped_data[department].append(count)

print("Shuffle and Sort Output:")
print(grouped_data)
```

```python
def reducer(department, counts):
    return (department, sum(counts))

print("Final Output:")

for department, counts in sorted(grouped_data.items()):
    print(reducer(department, counts))
```

```python
text = """Data science is evolving rapidly, and many students
aspire to learn data science. As data becomes more valuable,
the demand for data science skills continues to rise."""
```

```python
def mapper(line):
    words = line.lower().replace(',', '').replace('.', '').split()
    return [(word, 1) for word in words]

mapped_data = []

for line in text.splitlines():
    mapped_data.extend(mapper(line))

print("Mapper Output:")
print(mapped_data)
```

```python
grouped_data = {}

for word, count in mapped_data:
    if word not in grouped_data:
        grouped_data[word] = []
    grouped_data[word].append(count)

print("Shuffle and Sort Output:")
print(grouped_data
     )
```

```python
def reducer(word, counts):
    total = sum(counts)

    if total > 1:
        return (word, total)

print("Repeated Words:")

for word, counts in sorted(grouped_data.items()):
    result = reducer(word, counts)

    if result is not None:
        print(result)
```
